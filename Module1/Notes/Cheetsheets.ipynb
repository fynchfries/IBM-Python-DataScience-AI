# Python Cheatsheet

## Core Concepts

### Comments

A comment is text ignored by the Python interpreter when the program runs. Use `#` for a single-line comment.

```python
# This is a comment
```

### Variable Assignment

Variable assignment stores a value under a name.

**Syntax:**

```python
variable_name = value
```

**Example:**

```python
name = "John"  # Assigns "John" to name
x = 5           # Assigns 5 to x
```

### Data Types

Common Python data types include:

- **Integer (`int`)**: a whole number, such as `7`.
- **Float (`float`)**: a number with a decimal part, such as `12.4`.
- **Boolean (`bool`)**: a truth value, either `True` or `False`.
- **String (`str`)**: text enclosed in quotes, such as `"John"`.

```python
x = 7
price = 12.4
is_valid = True
is_complete = False
first_name = "John"
```

### Python Operators

Arithmetic operators perform calculations on values:

| Operator | Definition |
| --- | --- |
| `+` | Adds two values. |
| `-` | Subtracts the second value from the first. |
| `*` | Multiplies two values. |
| `/` | Divides the first value by the second and returns a float. |
| `//` | Floor-divides the first value by the second, rounding down to an integer result. |
| `%` | Returns the remainder after division. |

```python
x = 9
y = 4

result_add = x + y    # 13
result_sub = x - y    # 5
result_mul = x * y    # 36
result_div = x / y    # 2.25
result_fdiv = x // y  # 2
result_mod = x % y    # 1
```

## Strings

### Concatenation

Concatenation combines strings. Use the `+` operator.

**Syntax:**

```python
concatenated_string = string1 + string2
```

**Example:**

```python
result = "Hello" + " John"
```

### Indexing

Indexing accesses the character at a specific position in a string. Python indexes start at `0`.

```python
my_string = "Hello"
char = my_string[0]  # "H"
```

### Slicing

Slicing extracts part of a string. The start index is included, and the end index is excluded.

**Syntax:**

```python
substring = string_name[start:end]
```

**Example:**

```python
my_string = "Hello"
substring = my_string[0:5]  # "Hello"
```

## Built-in Functions

### `len()`

Returns the number of items in a sequence, including the number of characters in a string.

**Syntax:**

```python
len(string_name)
```

**Example:**

```python
my_string = "Hello"
length = len(my_string)  # 5
```

### `print()`

Displays the values passed to it.

```python
print("Hello, world")
print(2 + 3)
```

## String Methods

String methods operate on strings and return a result; they do not change the original string.

### `.lower()`

Returns a lowercase version of the string.

```python
my_string = "Hello"
lowercase_text = my_string.lower()  # "hello"
```

### `.upper()`

Returns an uppercase version of the string.

```python
my_string = "Hello"
uppercase_text = my_string.upper()  # "HELLO"
```

### `.replace()`

Returns a copy of the string with occurrences of one substring replaced by another.

```python
my_string = "Hello"
new_text = my_string.replace("Hello", "Hi")  # "Hi"
```

### `.split()`

Splits a string into a list of substrings using a delimiter. If no delimiter is provided, whitespace is used.

```python
my_string = "Hello, world"
split_text = my_string.split(",")  # ["Hello", " world"]
```

### `.strip()`

Returns a copy of the string with leading and trailing whitespace removed.

```python
my_string = "  Hello  "
trimmed = my_string.strip()  # "Hello"
```
